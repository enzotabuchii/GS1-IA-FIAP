## Sistema de Controle

```python
import random
import ollama


# Função para gerar dados simulados da missão
def gerar_dados_missao():
    # Simulando variações de cenários (Normal, Crítico de Energia, Crítico de Temperatura)
    cenario = random.choice(["normal", "alerta_temp", "alerta_energia"])

    if cenario == "normal":
        temperatura = random.randint(15, 40)  # °C
        energia = random.randint(70, 100)  # %
        comunicacao = "Estável"
    elif cenario == "alerta_temp":
        temperatura = random.randint(85, 110)  # °C (Crítico)
        energia = random.randint(40, 69)  # %
        comunicacao = "Instável"
    else:
        temperatura = random.randint(20, 35)  # °C
        energia = random.randint(5, 19)  # % (Crítico)
        comunicacao = "Estável"

    return temperatura, energia, comunicacao


# Estrutura lógica de tomada de decisão e alertas automáticos
def verificar_alertas_logicos(temperatura, energia, comunicacao):
    alertas = []
    acoes_automatizadas = []

    # Lógica para Temperatura
    if temperatura > 80:
        alertas.append("PERIGO: Superaquecimento nos módulos!")
        acoes_automatizadas.append(
            "[AÇÃO AUTOMÁTICA] Ativando sistema de resfriamento líquido criogênico."
        )

    # Lógica para Energia
    if energia < 20:
        alertas.append("PERIGO: Nível crítico de bateria!")
        acoes_automatizadas.append(
            "[AÇÃO AUTOMÁTICA] Desligando sistemas não-essenciais. Modo de Economia Ativo."
        )

    # Lógica para Comunicação
    if comunicacao == "Instável":
        alertas.append("ALERTA: Oscilação no sinal de rede.")
        acoes_automatizadas.append(
            "[AÇÃO AUTOMÁTICA] Redirecionando antenas para o satélite reserva."
        )

    # Caso esteja tudo bem
    if not alertas:
        alertas.append("Todos os parâmetros operando dentro da normalidade.")
        acoes_automatizadas.append(
            "[AÇÃO AUTOMÁTICA] Mantendo rotina padrão de monitoramento."
        )

    return alertas, acoes_automatizadas


# Integração com a IA Genérica (Ollama/Llama)
def obter_analise_ia(temperatura, energia, comunicacao):
    system_prompt = (
        "Você é o sistema de Inteligência Artificial do Controle de Missão Espacial (Mission Control AI). "
        "Sua função é analisar os dados de telemetria recebidos dos sensores, diagnosticar a gravidade da situação "
        "em um parágrafo curto e direto, e emitir uma diretriz ou previsão técnica para a tripulação."
    )

    user_prompt = f"Telemetria Atual -> Temperatura: {temperatura}°C | Energia: {energia}% | Comunicação: {comunicacao}. Avalie o status atual da missão."

    try:
        resposta = ollama.chat(
            model="llama3.2:1b",
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt},
            ],
        )
        return resposta["message"]["content"]
    except Exception as e:
        return f"Erro ao conectar com a IA: {e}"


# Interface visual organizada no Terminal / Output
def exibir_painel_controle(
    temperatura, energia, comunicacao, alertas, acoes, analise_ia
):
    print("=" * 64)
    print("==== MISSION CONTROL AI - STATUS ATUAL ====")
    print("=" * 64)
    print(f"Temperatura dos Módulos : {temperatura}°C")
    print(f"Nível de Energia        : {energia}%")
    print(f"Status da Comunicação   : {comunicacao}")
    print("-" * 60)

    print("ALERTAS DO SISTEMA:")
    for alerta in alertas:
        print(f"   {alerta}")
    print("=" * 60)

    print("PROTOCOLOS EXECUTADOS:")
    for acao in acoes:
        print(f"   {acao}")
    print("=" * 60)

    print("ANÁLISE PREDITIVA DA IA (Llama 3.2):")
    print(analise_ia)
    print("=" * 60)
    print("\n")


# Simulando uma rodada de checagem de dados
temp, energ, com = gerar_dados_missao()
lista_alertas, lista_acoes = verificar_alertas_logicos(temp, energ, com)
analise_da_ia = obter_analise_ia(temp, energ, com)

# Painel final
exibir_painel_controle(
    temp, energ, com, lista_alertas, lista_acoes, analise_da_ia
)
```
